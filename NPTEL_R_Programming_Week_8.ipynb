## Lecture 31 — Operations on Lists
Based on the NPTEL examples on merging, converting, appending, removing, and extracting from lists.

```r
# Create two lists
list1 <- list(1, 2, 3)
list2 <- list("water", "juice", "lemonade")

list1
list2
```

### Merging lists

```r
# Merge two lists
list12 <- c(list1, list2)

list12
```

### Converting a list to a vector with `unlist()`

```r
# Convert lists to vectors
unlist(list1)
unlist(list2)

# Compare modes
mode(list1)
mode(unlist(list1))
```

### Appending to a list

```r
# Append an item at the end
append(list1, 100)
append(list2, "coffee")
```

### Appending after a particular position

```r
# Insert after position 2
append(list1, 100, after = 2)
append(list2, "coffee", after = 2)
```

### Removing an item from a list

```r
# Remove the second item from list1
list1[-2]

# Remove the first item from list2
list2[-1]
```

### Extracting items from a list

```r
# Create longer lists
list1 <- list(1, 2, 3, 4, 5, 6)
list2 <- list("water", "juice", "lemonade", "tea", "coffee", "milk")

# Extract a range
list1[2:4]
list2[2:4]

# Extract selected positions
list1[c(1, 3, 5)]
list2[c(1, 3, 5)]
```

## Lecture 32 — Vector Indexing

```r
# Create a vector
x <- 1:10
x
```

### Indexing using logical conditions

```r
# Values greater than 5
x[x > 5]

# Even values
x[x %% 2 == 0]

# Odd values
x[x %% 2 == 1]
```

### Vector with missing observations

```r
x <- 1:10
x[5] <- NA
x

# Remove missing values
y <- x[!is.na(x)]
y

# Mean with NA
mean(x)

# Mean after removing NA
mean(y)
```

### Indexing with negative integers

```r
x <- 1:10

# Exclude positions 1 through 5
x[-(1:5)]

# Equivalent positive-index form
x[6:10]
```

### Named elements

```r
# A list with named elements
z <- list(a1 = 1, a2 = "c", a3 = 1:3)

z
names(z)
```

### Changing an element name

```r
z <- list(a1 = 1, a2 = "c", a3 = 1:3)

names(z)[3] <- "c2"
z
```

### Named vector indexing

```r
x <- c(water = 1, juice = 2, lemonade = 3)

names(x)
x["juice"]
```

### Empty index

```r
x <- 1:10

x
x[]
```

### Mixed-mode list

```r
# Lists can contain different data types
ab <- list(1, 2, 3, "X", "Y", "Z")

# Give the list matrix-like dimensions
dim(ab) <- c(2, 3)

print(ab)
mode(ab)
```

## Lecture 33 — Factors

### Categorical variables

Factors represent categorical variables and are used as grouping indicators.

Examples from the lecture:
- Gender: Male, Female
- Performance: Excellent, Good, Average, Bad
- Colours: Red, Blue, Green

### Creating a factor

```r
# Character vector representing categories
colours <- c("Red", "Green", "Green", "Blue", "Red")

# Convert to a factor
colour_factor <- factor(colours)

colour_factor
levels(colour_factor)
```

### Factor with specified levels and labels

```r
# Die outcomes
y <- c(1, 4, 3, 5, 4, 2, 4)

possible.dieface <- c(1, 2, 3, 4, 5, 6)

labels.dieface <- c(
  "one", "two", "three", "four", "five", "six"
)

facy <- factor(
  y,
  levels = possible.dieface,
  labels = labels.dieface
)

facy
levels(facy)
```

### Character vector versus factor

```r
y
facy

# Display the structure
str(y)
str(facy)
```

## Lecture 34 — Factors: Class and Unclass

### `as.factor()`

```r
x <- c(3, 4, 5, 6, 1, 2, 3, 3, 4, 4, 5, 6)

x
y <- as.factor(x)

y
levels(y)
```

### Factor levels for character data

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water")
)

x
levels(x)
```

### `class()`

```r
# Check classes of different objects
class(9)
class("9")
class(print)

x <- matrix(nrow = 2, ncol = 2, data = 1:4)
class(x)
```

### `unclass()`

```r
brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)

brands_fac
unclass(brands_fac)
```

### Using factor codes to index another vector

```r
colours <- c("blue", "green", "red")

brands <- c("A", "A", "B", "B", "B", "B", "C")
brands_fac <- factor(brands)

# The numeric codes identify factor levels
unclass(brands_fac)

# Use the codes as indexes
colours[unclass(brands_fac)]
```

### Factor level order

```r
x <- factor(
  c("lemonade", "lemonade", "juice", "lemonade", "water"),
  levels = c("water", "juice", "lemonade")
)

x
unclass(x)
levels(x)
```

### Ordered factor

```r
income <- ordered(
  c("high", "high", "low", "medium", "medium"),
  levels = c("low", "medium", "high")
)

income
levels(income)
unclass(income)
```

## Lecture 35 — Strings: Display and Formatting

Important commands covered in the lecture:
- `print()`
- `format()`
- `cat()`
- `paste()`

This notebook focuses on the examples shown in the supplied lecture notes.

### `print()` with numbers

```r
print(sqrt(2))
print(sqrt(2), digits = 5)
print(sqrt(2), digits = 10)
```

### `print()` with character vectors

```r
print("apple")
print(c("apple", "banana"))

# Mixed values in a vector become character data
print(c("apple", "banana", 6, 10))
```

### `format()`

```r
# Format a numeric value
format(0.5, digits = 10, nsmall = 15)

# Print the formatted result
print(format(0.5, digits = 10, nsmall = 15))
```

### `format()` with width and justification

```r
x <- c("A", "BB", "CCC", "DDDD")

format(x, width = 7, justify = "centre")
format(x, width = 7, justify = "left")
format(x, width = 7, justify = "right")
format(x, width = 7, justify = "none")
```

### Printing a matrix

```r
x <- matrix(
  nrow = 3,
  ncol = 2,
  data = 1:6,
  byrow = TRUE
)

print(x)
```

### Formatting large quantities

```r
format(1234567, big.mark = ",")
format(12345678, big.mark = ",")
format(123456789, big.mark = ",")
```

### Formatting large quantities with spaces

```r
format(123456789, big.mark = " ")
```
