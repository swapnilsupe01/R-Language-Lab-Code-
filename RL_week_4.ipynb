Matrix Access & Mathematical Operations

## Access rows, columns and submatrices


```r
x <- matrix(nrow = 5, ncol = 3, byrow = TRUE, data = 1:15)
x
```

```r

# Access rows, columns and submatrices
x[3,]
x[,2]
x[4:5, 2:3]
x[c(1,4), c(1,3)]
```

```r
# Scalar addition
x <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
x
x + 5
```

```r
# Scalar subtraction
x - 5
```

```r
# Scalar multiplication
5 * x
```

```r
# Scalar division
x / 2
```

```r
# Addition and subtraction of matrices
x <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
y <- matrix(nrow = 4, ncol = 2, data = 11:18, byrow = TRUE)
x
y
x + y
x - y
```

```r
# More scalar examples
4 * x
x + 4 * x
4 * x - x
```

```r
# Matrix multiplication
x <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
y <- matrix(nrow = 2, ncol = 4, data = 11:18, byrow = TRUE)
x %*% y
y %*% x
```

```r
# X'X and XX'
x <- matrix(nrow = 4, ncol = 2, data = 1:8, byrow = TRUE)
t(x) %*% x
x %*% t(x)
```

```r
# Cross product
crossprod(x)
```

```r
# Row-wise and column-wise concatenation
x <- matrix(nrow = 3, ncol = 2, data = 1:6, byrow = TRUE)
y <- matrix(nrow = 3, ncol = 2, data = 11:16, byrow = TRUE)
rbind(x, y)
cbind(x, y)
```

```r
# Inverse of a matrix
y <- matrix(nrow = 2, ncol = 2, byrow = TRUE,
            data = c(84, 100, 100, 120))
solve(y)
```

```r
# Eigenvalues and eigenvectors
eigen(y)
```

Logical & Relational Operators

## Comparison operators:


```r
# Comparison operators:
# >  >=  <  <=  ==  !=  !
# &  &&  |  ||  xor()
# isTRUE()  isFALSE()

# OR: || uses only the first element; | is element-wise
x <- 8
(x < 10) || (x < 2)
```

```r
x <- 18
(x < 10) || (x < 2)
```

```r
x <- c(8, 18)
(x < 10) | (x < 2)
(x < 10) | (x < 2)
```

```r
# AND: && uses only the first element; & is element-wise
x <- 5
(x < 10) && (x > 2)
```

```r
x <- 15
(x < 10) && (x > 2)
```

```r
x <- c(8, 18)
(x < 10) & (x > 2)
(x < 10) & (x > 2)
```

```r

# Filtering using logical conditions
x <- 1:6
(x > 2) & (x < 5)
x[(x > 2) & (x < 5)]
```

```r
(x > 2) | (x < 5)
x[(x > 2) | (x < 5)]
```

```r
# Longer-form logical operation on the first vector element
x <- 1:6
(x > 2) & (x < 5)
(x[1] > 2) & (x[1] < 5)
```

```r
# Truth table examples
x <- TRUE
y <- FALSE
x & y
x | y
!x
```

```r
# Logical values and is.logical()
x <- 5

Logical1 <- (x > 2)
Logical1
is.logical(Logical1)
```

```r
Logical1 <- (x > 2)
Logical1
is.logical(Logical1)
```

```r
Logical2 <- (x < 10)
Logical2
is.logical(Logical2)
```

```r
Logical3 <- (x != 5)
Logical3
is.logical(Logical3)
```

```r
Logical4 <- (2 * x > 11)
Logical4
is.logical(Logical4)
```

```r
Logical5 <- (3 * x < 20)
Logical5
is.logical(Logical5)
```

```r
# Relational examples
8 > 7
7 < 5
7 > 7
7 >= 7
8 < 8
8 <= 8
```

```r
8 != 9
9 != 9
7 == 7
7 == 8
```

```r
x <- TRUE
!x
```

```r
# Vector comparisons
x <- c(1, 2, 3)
y <- c(4, 5, 6)
x > y
x < y
x != y
x == y
```

```r

# isTRUE / isFALSE
isTRUE(8 < 6)
isTRUE(8 > 6)
isFALSE(5 < 8)
isFALSE(5 > 8)
```
